```python
df_analysis = df_train.merge(
    df,
    on = 'msno',
    how='left'
)
```

```python
df_analysis.head()
```

```python
#все ли пользователи из df_train нашли данные в members

df_analysis.isna().sum()
```

```python
missing_members = df_analysis['city'].isna().sum()
missing_pct = missing_members / len(df_analysis) * 100

print(f'Нет данных в members: {missing_members:,} пользователей ({missing_pct:.1f}%)')
```

```python
df_analysis['has_member_data'] = df_analysis['city'].notna().astype(int)
```

```python
#среди пользователей без данных в members ушло примерно 5.3%;
#среди пользователей с данными в members ушло примерно 9.5%.

df_analysis.groupby('has_member_data')['is_churn'].mean() * 100
```

```python
#смотрим churn по признакам из members

df_analysis.groupby('gender')['is_churn'].mean() * 100
```

```python
df_analysis['gender'].value_counts(dropna=False)
```

```python
df_analysis.groupby('registered_via')['is_churn'].agg(['count', 'mean']).head()
```

```python
df_analysis.groupby('city')['is_churn'].agg(['count', 'mean']).head()
```

```python
bins = [0, 18, 25, 35, 45, 55, 65, 100]
labels = ['<18', '18-24', '25-34', '35-44', '45-54', '55-64', '65+']

df_analysis['age_group'] = pd.cut(
    df_analysis['bd'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
df_analysis.groupby('age_group', observed=False)['is_churn'].agg(['count', 'mean'])
```

```python
#проверим стаж пользователя

#сколько дней прошло с регистр
reference_date = pd.Timestamp('2017-04-30')
df_analysis['account_age_days'] = (
    reference_date - df_analysis['registration_init_time']
).dt.days
```

```python
#переведем дни в годы
df_analysis['account_age_years'] = (
    df_analysis['account_age_days'] / 365.25
)
```

```python
#сделаем группы по стажу
bins = [0, 1, 2, 3, 5, 10, 20]
labels = ['<1 year', '1-2', '2-3', '3-5', '5-10', '10+']

df_analysis['account_age_group'] = pd.cut(
    df_analysis['account_age_years'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
account_churn = df_analysis.groupby(
    'account_age_group',
    observed=False
)['is_churn'].agg(['count', 'mean'])

account_churn['mean'] = account_churn['mean'] * 100

account_churn
```

```python
#кол-во транзакций на пользователя

transactions_agg = df_transactions.groupby('msno').agg(
    transactions_count = ('msno', 'size')
).reset_index()
transactions_agg.head()
```

```python
#кто сделал больше всего транзакций
transactions_agg.sort_values(
    'transactions_count',
    ascending=False
).head(10)
```

```python
#присоед transactions_agg к df_analysis

df_analysis = df_analysis.merge(
    transactions_agg,
    on='msno',
    how = 'left'
)
```

```python
#36452 - пользователи, которые не делали транзакций
df_analysis['transactions_count'].isna().sum()
```

```python
df_analysis['transactions_count'] = (df_analysis['transactions_count'].fillna(0))
```

```python
df_analysis.groupby('transactions_count')['is_churn'].agg(['count', 'mean']).head(10)
```

```python
last_transactions = (
    df_transactions.sort_values('transaction_date').groupby('msno').tail(1)
)
```

```python
last_transactions = last_transactions[
    [
        'msno',
        'payment_plan_days',
        'plan_list_price',
        'actual_amount_paid',
        'is_auto_renew',
        'is_cancel',
        'transaction_date',
        'membership_expire_date'
    ]
]
```

```python
df_analysis = df_analysis.merge(
    last_transactions,
    on='msno',
    how='left'
)
```

```python
#Пользователи без активного автопродления имеют значительно более высокий churn отток
df_analysis.groupby('is_auto_renew')['is_churn'].agg(['count', 'mean'])
```

```python
#is_cancel 1 - отменил автопродление
df_analysis.groupby('is_cancel')['is_churn'].agg(['count', 'mean'])
```

```python
#Среди пользователей с включённым автопродлением какой процент отменяет подписку
pd.crosstab(
    df_analysis['is_auto_renew'],
    df_analysis['is_cancel'],
    normalize='index'
) * 100
```

```python
#рассмотрим длительность подписки

df_analysis.groupby('payment_plan_days')['is_churn'].agg(['count', 'mean']).sort_values('count', ascending=False).head(15)
```

```python
bins = [0, 31, 91, 181, 366, 1000]
labels = ['~1 month', '2-3 months', '4-6 months', '7-12 months', '12+ months']

df_analysis['plan_group'] = pd.cut(
    df_analysis['payment_plan_days'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
plan_churn = df_analysis.groupby(
    'plan_group',
    observed=False
)['is_churn'].agg(['count', 'mean'])

plan_churn['mean'] = plan_churn['mean'] * 100
plan_churn
```

```python
df_analysis['has_discount'] = (
    df_analysis['actual_amount_paid'] < df_analysis['plan_list_price']
).astype(int)
```

```python
#заплатил меньше прайса - 1, полную 0
discount_churn = df_analysis.groupby('has_discount')['is_churn'].agg(['count', 'mean'])
discount_churn['mean'] = discount_churn['mean'] * 100
discount_churn
```

```python
df_analysis['actual_amount_paid'].describe()
```

```python
df_analysis['payment_group'] = pd.cut(
    df_analysis['actual_amount_paid'],
    bins=[-1, 99, 149, 2000],
    labels=['0-99', '100-149', '150+']
)
```

```python
df_analysis.groupby('payment_group', observed=False)['is_churn'].agg(['count', 'mean'])
```

```python
#как связан размер платежа и длителььнось тарифа
pd.crosstab(
    df_analysis['payment_group'],
    df_analysis['plan_group'],
    normalize='index'
) * 100
```

```python
df_analysis.groupby('payment_group')['payment_plan_days'].mean()
```

```python
#Чем дольше пользователь ничего не делал, тем логичнее ожидать более высокий churn.

reference_date = pd.Timestamp('2017-04-30')

df_analysis['days_since_last_transaction'] = (
    reference_date - df_analysis['transaction_date']
).dt.days
```

```python
df_analysis['days_since_last_transaction'].describe()
```

```python
bins = [-1, 7, 30, 60, 90, 180, 10000]
labels = [
    '0-7 days',
    '8-30 days',
    '31-60 days',
    '61-90 days',
    '91-180 days',
    '180+ days'
]

df_analysis['recency_group'] = pd.cut(
    df_analysis['days_since_last_transaction'],
    bins=bins,
    labels=labels
)
```

```python
recency_churn = df_analysis.groupby(
    'recency_group',
    observed=False
)['is_churn'].agg(['count', 'mean'])

recency_churn['mean'] = recency_churn['mean'] * 100
recency_churn
```

```python
df_analysis.to_csv('customer_retention_ready.csv', index=False)
```
